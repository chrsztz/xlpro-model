Tensor operation

```python
tensor = torch.tensor([1,2,3])
tensor = torch.sub(tensor,1)
torch.mul(tensor,10)
```

Transpose a matrix

```python
tensor1 = torch.rand(3,2)
print(tensor1.shape,tensor2.shape)
tensor1 = tensor1.T #change to (2,3) matrix
torch.mm(tensor1,tensor2)
```

Tensor aggregation: finding min, max, mean, sum...

```python
x = torch.arange(0,100,10)
x.min(),torch.min(x),x.max(),torch.max(x),x.sum(),torch.sum(x),x.type(torch.float32).mean(),torch.mean(x.type(torch.float32)) #mean function require float datatype
```

```python
#find the position
x.argmin(),x.argmax(),x[x.argmin()],x[x.argmax()]
```

Reshaping, stacking, squeezing and unsqueezing tensors

```python
x = torch.arange(1.,10.)
x, x.shape
```

```python
# add a dimension- 元素数量须符合
x_reshaped = x.reshape(3,3)
x_reshaped,x_reshaped.shape
```

```python
# change the view
z = x.view(1,9)
z, z.shape
```

* name.view与name.reshape的区别: name.view共享内存，name.reshape进行赋值

```python
z[:,0] = 5
z,x
```
