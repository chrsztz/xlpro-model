### 张量(Tensor)操作

#### 创建张量

```python
import torch
# 标量（0维张量）
scalar = torch.tensor(3.14)
# 向量（1维张量）
vector = torch.tensor([1.0, 2.0, 3.0])
# 矩阵（2维张量）
matrix = torch.tensor([[1, 2], [3, 4]])
# 3维张量
tensor_3d = torch.rand(2, 3, 4)

print("Scalar:", scalar)
print("Vector:", vector)
print("Matrix:", matrix)
print("3D Tensor:", tensor_3d)
```

#### 张量运算

```python
a = torch.tensor([2.0, 3.0])
b = torch.tensor([4.0, 1.0])

# 基本运算
print("a + b:", a + b)
print("a * b:", a * b)
print("矩阵乘法:", torch.matmul(a.unsqueeze(0), b.unsqueeze(1)))  # 将向量调整为矩阵形状
```

#### 自动求导 (Autograd)

```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2 + 3 * x + 1  # 函数表达式
y.backward()  # 自动求导
print("x 的梯度:", x.grad)  # dy/dx = 2x + 3
```

### 线性回归

#### 数据生成

```python
import torch
import matplotlib.pyplot as plt

# 生成数据
x = torch.linspace(0, 10, 100).unsqueeze(1)  # x: 100 个点，形状 (100, 1)
y = 2 * x + 1 + torch.randn_like(x) * 2  # y: 增加随机噪声

plt.scatter(x.numpy(), y.numpy())
plt.title("Generated Data")
plt.show()
```

#### 定义模型

```python
import torch.nn as nn

class LinearRegressionModel(nn.Module):
    def __init__(self):
        super(LinearRegressionModel, self).__init__()
        self.linear = nn.Linear(1, 1)  # 定义线性层

    def forward(self, x):
        return self.linear(x)

model = LinearRegressionModel()
print(model)
```

#### 损失函数与优化器

```python
criterion = nn.MSELoss()  # 均方误差
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)  # 随机梯度下降
```

#### 模型训练

```python
for epoch in range(100):  # 训练 100 个周期
    optimizer.zero_grad()         # 清空梯度
    predictions = model(x)        # 前向传播
    loss = criterion(predictions, y)  # 计算损失
    loss.backward()               # 反向传播
    optimizer.step()              # 更新参数

    if (epoch + 1) % 10 == 0:  # 每 10 个周期打印一次
        print(f'Epoch [{epoch+1}/100], Loss: {loss.item():.4f}')
```

#### 模型评估

```python
with torch.no_grad():
    predictions = model(x)
    plt.scatter(x.numpy(), y.numpy(), label="Original Data")
    plt.plot(x.numpy(), predictions.numpy(), color="red", label="Fitted Line")
    plt.legend()
    plt.title("Model Fit")
    plt.show()
```
